## Data Cleaning

```python
# ============================================================
#                      DATA CLEANING
# ============================================================

print("=" * 60)
print("             BEFORE DATA CLEANING")
print("=" * 60)

print("\nShape of Dataset :", df.shape)

print("\nMissing Values in Each Column:\n")
print(df.isnull().sum())

print("\nDuplicate Records :", df.duplicated().sum())

# ============================================================
#              REMOVING DUPLICATE RECORDS
# ============================================================

duplicate_count = df.duplicated().sum()

print("Duplicate Rows Found :", duplicate_count)

df.drop_duplicates(inplace=True)

print("Duplicate Rows Removed Successfully!")

print("Current Shape :", df.shape)

# ============================================================
#          CHECKING MISSING VALUE PERCENTAGE
# ============================================================

missing = df.isnull().sum()

missing_percentage = (missing / len(df)) * 100

missing_df = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percentage
})

missing_df = missing_df.sort_values(
    by="Percentage",
    ascending=False
)

display(missing_df)

# ============================================================
#            VISUALIZATION OF MISSING VALUES
# ============================================================

plt.figure(figsize=(10, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    cmap="viridis"
)

plt.title("Missing Values Heatmap")

plt.show()

# ============================================================
#          DROPPING SOCIETY COLUMN
# ============================================================

print("Columns Before Dropping:")
print(df.columns)

df.drop("society", axis=1, inplace=True)

print("\nColumns After Dropping:")
print(df.columns)

# ============================================================
#           HANDLING MISSING VALUES
# ============================================================

# Fill Bathroom values with Median
df["bath"] = df["bath"].fillna(
    df["bath"].median()
)

# Fill Balcony values with Median
df["balcony"] = df["balcony"].fillna(
    df["balcony"].median()
)

print("Bath Missing Values :", df["bath"].isnull().sum())
print("Balcony Missing Values :", df["balcony"].isnull().sum())

# ============================================================
#         DROPPING REMAINING NULL VALUES
# ============================================================

df.dropna(inplace=True)

print("Dataset Shape After Removing Null Values:")
print(df.shape)

print("\nRemaining Missing Values:\n")
print(df.isnull().sum())

# ============================================================
#             FINAL CLEANED DATASET
# ============================================================

print("=" * 60)
print("              CLEANING COMPLETED")
print("=" * 60)

print("\nFinal Shape :", df.shape)

print("\nFirst 5 Records:")
display(df.head())

print("\nLast 5 Records:")
display(df.tail())
```

## Feature Engineering

```python
# ============================================================
#              FEATURE ENGINEERING - BHK EXTRACTION
# ============================================================

print("=" * 60)
print("        EXTRACTING BHK FROM SIZE COLUMN")
print("=" * 60)

# Display sample values
print("\nSample Values from 'size' Column:")
print(df["size"].head(10))

# Create new BHK column
df["bhk"] = df["size"].apply(lambda x: int(str(x).split(" ")[0]))

print("\nBHK Column Created Successfully!")

print("\nFirst 10 Values of BHK:")
print(df["bhk"].head(10))

print("\nUnique BHK Values:")
print(sorted(df["bhk"].unique()))

# ============================================================
#                 VISUALIZING BHK DISTRIBUTION
# ============================================================

plt.figure(figsize=(10,6))

sns.countplot(
    x="bhk",
    data=df,
    order=sorted(df["bhk"].unique())
)

plt.title("Distribution of BHK")
plt.xlabel("Number of Bedrooms (BHK)")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()

# ============================================================
#            CONVERTING TOTAL SQFT TO NUMERIC
# ============================================================

print("=" * 60)
print("      CHECKING SAMPLE VALUES OF TOTAL SQFT")
print("=" * 60)

print(df["total_sqft"].head(15))

# ============================================================
#          FUNCTION TO CONVERT TOTAL SQFT VALUES
# ============================================================

de  f convert_sqft_to_num(x):

    try:
        if "-" in str(x):
            tokens = x.split("-")
            return (
                float(tokens[0]) +
                float(tokens[1])
            ) / 2
        else:
            return float(x)

    except:
        return np.nan

# ============================================================
#           APPLYING CONVERSION FUNCTION
# ============================================================

df["total_sqft"] = df["total_sqft"].apply(convert_sqft_to_num)
print("Conversion Completed Successfully!")
display(df.head())

# ============================================================
#          CHECKING NULL VALUES AFTER CONVERSION
# ============================================================

print(df["total_sqft"].isnull().sum())

# ============================================================
#             REMOVING INVALID RECORDS
# ============================================================

before_rows = df.shape[0]

df.dropna(subset=["total_sqft"], inplace=True)

after_rows = df.shape[0]

print("Rows Before :", before_rows)
print("Rows After  :", after_rows)
print("Rows Removed:", before_rows - after_rows)

# ============================================================
#         DISTRIBUTION OF TOTAL SQUARE FEET
# ============================================================

plt.figure(figsize=(10,6))

sns.histplot(
    df["total_sqft"],
    bins=50,
    kde=True
)

plt.title("Distribution of Total Square Feet")
plt.xlabel("Total Square Feet")
plt.ylabel("Frequency")

plt.show()

# ============================================================
#           CREATING PRICE PER SQUARE FEET
# ============================================================

df["price_per_sqft"] = (
    df["price"] * 100000
) / df["total_sqft"]

print("New Feature Added Successfully!")

display(
    df[
        [
            "total_sqft",
            "price",
            "price_per_sqft"
        ]
    ].head()
)

# ============================================================
#             DROPPING SIZE COLUMN
# ============================================================

df.drop(
    "size",
    axis=1,
    inplace=True
)

print("Size Column Removed Successfully!")
print("\nRemaining Columns:\n")
print(df.columns)

# ============================================================
#        SAVE CLEANED DATASET
# ============================================================

df.to_csv("Bengaluru_House_Cleaned.csv", index=False)
print("Cleaned Dataset Saved Successfully!")
```

## Encoding Categorial Columns

```python
# ============================================================
#          CHECKING CATEGORICAL COLUMNS
# ============================================================

print("=" * 60)
print("         CATEGORICAL COLUMNS IN DATASET")
print("=" * 60)

categorical_columns = df.select_dtypes(include=["object"]).columns
print(categorical_columns)
print("\n")

for column in categorical_columns:
    print("-" * 40)
    print("Column Name :", column)
    print("Unique Values :", df[column].nunique())
    print(df[column].value_counts().head())
    print()

# ============================================================
#          REDUCING RARE LOCATIONS
# ============================================================

location_count = df["location"].value_counts()

print(location_count.head(20))

# Strip unwanted spaces
df["location"] = df["location"].apply(lambda x: x.strip())

# Count again
location_count = df["location"].value_counts()

# Replace locations having less than 10 houses
location_count_less_than_10 = location_count[location_count <= 10]

print("\nRare Locations :", len(location_count_less_than_10))

df["location"] = df["location"].apply(
    lambda x: "other"
    if x in location_count_less_than_10
    else x
)

print("\nUpdated Unique Locations :", df["location"].nunique())

# ============================================================
#             ONE HOT ENCODING
# ============================================================

print("=" * 60)
print("      APPLYING ONE HOT ENCODING")
print("=" * 60)

categorical_cols = [
    "area_type",
    "availability",
    "location"
]

df = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)

print("Encoding Completed Successfully!\n")
print("Updated Shape :", df.shape)
display(df.head())

# ============================================================
#        SAVE ENCODED DATASET
# ============================================================

df.to_csv("Bengaluru_House_Encoded.csv", index=False)
print("Encoded Dataset Saved Successfully!")

# ============================================================
#             VERIFYING DATA TYPES
# ============================================================

print(df.info())
```

## Correlation Analysis

```python
# ============================================================
#              CORRELATION ANALYSIS
# ============================================================

plt.figure(figsize=(14,10))

numeric_columns = [
    "total_sqft",
    "bath",
    "balcony",
    "bhk",
    "price",
    "price_per_sqft"
]

correlation_matrix = df[numeric_columns].corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()

# ============================================================
#      CORRELATION OF FEATURES WITH PRICE
# ============================================================

corr = correlation_matrix["price"].sort_values(ascending=False)
print(corr)
plt.figure(figsize=(8,5))

corr.plot(
    kind="bar",
    color="teal"
)

plt.title("Correlation of Features with Price")
plt.ylabel("Correlation Value")
plt.grid(axis="y")
plt.show()

# ============================================================
#      SCATTER PLOT : TOTAL SQFT VS PRICE
# ============================================================

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="total_sqft",
    y="price",
    alpha=0.6
)

plt.title("Total Square Feet vs House Price")
plt.xlabel("Total Square Feet")
plt.ylabel("Price (Lakhs)")
plt.show()

if "price_per_sqft" in df.columns:
    df.drop("price_per_sqft", axis=1, inplace=True)
```

## Train - Test Split

```python
# ============================================================
#                TRAIN - TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split

print("=" * 60)
print("            PREPARING TRAINING DATA")
print("=" * 60)

# Drop target leakage column if it exists
if "price_per_sqft" in df.columns:
    df.drop("price_per_sqft", axis=1, inplace=True)

# Define Features (X) and Target (y)
X = df.drop("price", axis=1)
y = df["price"]

X.to_csv("X_features.csv", index=False)
y.to_csv("y_target.csv", index=False)
print("X and y saved successfully!")

print("\nFeatures Shape :", X.shape)
print("Target Shape   :", y.shape)

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Data Shape")
print("-------------------")
print("X_train :", X_train.shape)
print("y_train :", y_train.shape)

print("\nTesting Data Shape")
print("-------------------")
print("X_test  :", X_test.shape)
print("y_test  :", y_test.shape)
```

## Data Modeling

```python
# ============================================================
#                IMPORTING ML MODELS
# ============================================================

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

print("Models Imported Successfully!")

# ============================================================
#             LINEAR REGRESSION MODEL
# ============================================================

linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
linear_pred = linear_model.predict(X_test)
print("Linear Regression Model Trained Successfully!")

# ============================================================
#           DECISION TREE REGRESSION MODEL
# ============================================================

decision_model = DecisionTreeRegressor(
    random_state=42
)
decision_model.fit(X_train, y_train)
decision_pred = decision_model.predict(X_test)
print("Decision Tree Model Trained Successfully!")

# ============================================================
#          RANDOM FOREST REGRESSION MODEL
# ============================================================

random_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

random_model.fit(X_train, y_train)
random_pred = random_model.predict(X_test)
print("Random Forest Model Trained Successfully!")
```

## Model Evaluation

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ============================================================
#               MODEL COMPARISON
# ============================================================

import numpy as np

models = {
    "Linear Regression": linear_pred,
    "Decision Tree": decision_pred,
    "Random Forest": random_pred
}

results = []

for name, pred in models.items():
    mae = mean_absolute_error(y_test, pred)
    mse = mean_squared_error(y_test, pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, pred)
    results.append([name, mae, rmse, r2])

comparison = pd.DataFrame(
    results,
    columns=[
        "Model",
        "MAE",
        "RMSE",
        "R2 Score"
    ]
)

comparison = comparison.sort_values(
    by="R2 Score",
    ascending=False
)

display(comparison)

# ============================================================
#               MODEL COMPARISON GRAPH
# ============================================================

plt.figure(figsize=(8,5))

sns.barplot(
    data=comparison,
    x="Model",
    y="R2 Score"
)

plt.title("Comparison of Regression Models")
plt.xlabel("Models")
plt.ylabel("R2 Score")
plt.show()

# ============================================================
#              SELECT BEST MODEL
# ============================================================

best_model_name = comparison.iloc[0]["Model"]
print("Best Performing Model :", best_model_name)

if best_model_name == "Linear Regression":
    final_model = linear_model
elif best_model_name == "Decision Tree":
    final_model = decision_model
else:
    final_model = random_model
print("Final Model Selected Successfully!")
```

## Export Final Model

```python
# ============================================================
#                EXPORT FINAL MODEL
# ============================================================

import joblib

joblib.dump(
    final_model,
    "House_Price_Prediction_Model.pkl"
)
joblib.dump(
    X.columns.tolist(),
    "model_columns.pkl"
)
print("Model Saved Successfully!")
```

```python
import customtkinter as ctk
import tkinter as tk
import pandas as pd
import numpy as np
import joblib

# Load Saved Model
model = joblib.load("House_Price_Prediction_Model.pkl")

# Load Feature Columns
model_columns = joblib.load("model_columns.pkl")

# APP THEME
ctk.set_appearance_mode("dark")
ctk.set_default_color_theme("blue")

app = ctk.CTk()
app.geometry("900x700")
app.title("Bengaluru House Price Prediction")
app.resizable(False, False)

# TITLE
title = ctk.CTkLabel(
    app,
    text="🏠 Bengaluru House Price Prediction",
    font=("Arial", 30, "bold")
)

title.pack(pady=20)

subtitle = ctk.CTkLabel(
    app,
    text="Predict House Prices using Machine Learning",
    font=("Arial", 18)
)

subtitle.pack(pady=5)

# ===========================
# INPUT SECTION
# ===========================

data = pd.read_csv("Bengaluru_House_Cleaned.csv")

location_list = sorted(data["location"].dropna().unique().tolist())

form_frame = ctk.CTkFrame(app, fg_color="transparent")
form_frame.pack(pady=20)

# -------- Area Type --------
ctk.CTkLabel(
    form_frame,
    text="Area Type",
    width=140,
    anchor="w",
    font=("Arial", 15, "bold")
).grid(row=0, column=0, padx=10, pady=8)

area_combo = ctk.CTkComboBox(
    form_frame,
    values=[
        "Super built-up Area",
        "Built-up Area",
        "Plot Area",
        "Carpet Area"
    ],
    width=320
)
area_combo.grid(row=0, column=1, pady=8)

# -------- Availability --------
ctk.CTkLabel(
    form_frame,
    text="Availability",
    width=140,
    anchor="w",
    font=("Arial", 15, "bold")
).grid(row=1, column=0, padx=10, pady=8)

availability_combo = ctk.CTkComboBox(
    form_frame,
    values=[
        "Ready To Move",
        "Immediate Possession",
        "Under Construction"
    ],
    width=320
)
availability_combo.grid(row=1, column=1, pady=8)

# -------- Location --------
ctk.CTkLabel(
    form_frame,
    text="Location",
    width=140,
    anchor="w",
    font=("Arial", 15, "bold")
).grid(row=2, column=0, padx=10, pady=8)

location_combo = ctk.CTkComboBox(
    form_frame,
    values=location_list,
    width=320
)
location_combo.grid(row=2, column=1, pady=8)

# -------- Total Sqft --------
ctk.CTkLabel(
    form_frame,
    text="Total Sqft",
    width=140,
    anchor="w",
    font=("Arial", 15, "bold")
).grid(row=3, column=0, padx=10, pady=8)

sqft_entry = ctk.CTkEntry(form_frame, width=320)
sqft_entry.grid(row=3, column=1, pady=8)

# -------- Bathrooms --------
ctk.CTkLabel(
    form_frame,
    text="Bathrooms",
    width=140,
    anchor="w",
    font=("Arial", 15, "bold")
).grid(row=4, column=0, padx=10, pady=8)

bath_entry = ctk.CTkEntry(form_frame, width=320)
bath_entry.grid(row=4, column=1, pady=8)

# -------- Balconies --------
ctk.CTkLabel(
    form_frame,
    text="Balconies",
    width=140,
    anchor="w",
    font=("Arial", 15, "bold")
).grid(row=5, column=0, padx=10, pady=8)

balcony_entry = ctk.CTkEntry(form_frame, width=320)
balcony_entry.grid(row=5, column=1, pady=8)

# -------- BHK --------
ctk.CTkLabel(
    form_frame,
    text="BHK",
    width=140,
    anchor="w",
    font=("Arial", 15, "bold")
).grid(row=6, column=0, padx=10, pady=8)

bhk_entry = ctk.CTkEntry(form_frame, width=320)
bhk_entry.grid(row=6, column=1, pady=8)

# PREDICTION FUNCTION
result_label = ctk.CTkLabel(
    app,
    text="",
    font=("Arial", 22, "bold"),
    text_color="lightgreen"
)

result_label.pack(pady=20)

def predict_price():

    try:

        sqft = float(sqft_entry.get())
        bath = float(bath_entry.get())
        balcony = float(balcony_entry.get())
        bhk = int(bhk_entry.get())

        # Create a zero-filled row with all training columns
        input_df = pd.DataFrame(
            np.zeros((1, len(model_columns))),
            columns=model_columns
        )

        # Fill numeric features
        if "total_sqft" in input_df.columns:
            input_df.loc[0, "total_sqft"] = sqft

        if "bath" in input_df.columns:
            input_df.loc[0, "bath"] = bath

        if "balcony" in input_df.columns:
            input_df.loc[0, "balcony"] = balcony

        if "bhk" in input_df.columns:
            input_df.loc[0, "bhk"] = bhk

        # Fill one-hot encoded categorical columns
        area_col = f"area_type_{area_combo.get().strip()}"
        if area_col in input_df.columns:
            input_df.loc[0, area_col] = 1

        avail_col = f"availability_{availability_combo.get().strip()}"
        if avail_col in input_df.columns:
            input_df.loc[0, avail_col] = 1

        location_col = f"location_{location_combo.get().strip()}"
        if location_col in input_df.columns:
            input_df.loc[0, location_col] = 1
    
        prediction = model.predict(input_df)[0]

        result_label.configure(
            text=f"💰 Estimated Price : ₹ {prediction:.2f} Lakhs"
        )

    except Exception as e:

        result_label.configure(
            text=f"Error: {e}"
        )

# BUTTONS
predict_button = ctk.CTkButton(
    app,
    text="🔍 Predict Price",
    command=predict_price,
    width=220,
    height=45,
    font=("Arial", 18, "bold")
)

predict_button.pack(pady=10)

def clear_fields():

    area_combo.set("Super built-up Area")
    availability_combo.set("Ready To Move")
    location_combo.set(location_list[0]) 
    sqft_entry.delete(0, tk.END)
    bath_entry.delete(0, tk.END)
    balcony_entry.delete(0, tk.END)
    bhk_entry.delete(0, tk.END)

    result_label.configure(text="")

clear_button = ctk.CTkButton(
    app,
    text="🔄 Clear",
    command=clear_fields,
    fg_color="red"
)

clear_button.pack(pady=5)

# RUN APP
app.mainloop()
```
